# **3. 인공지능 모델 학습**  
### <font color='red'>KNN 알고리즘</font>
* knn은 지도학습이지만 훈련되는 것이 없다. **단순히 데이터를 가지고 있는 것이 전부**  
* 새로운 데이터가 입력되면 주변 데이터와의 거리 계산(manhattan 1, Unclidean 2)으로 분류하기 때문

*   라이브러리 가져오기

````python
from sklearn.neighbors import KNeighborsClassifier
````

*   학습용 데이터 생성하기

````python
fish_x=fish[['Length1', 'Weight']]
fish_y=fish['target']
````

*   모델 학습

````python
kn=KNeighborsClassifier()
kn.fit(fish_x, fish_y)
````

*   모델 평가

````python
kn.score(fish_x, fish_y)
````

*   데이터 결과 예측

````python
kn.predict([[30,600]])
````

````python
plt.scatter(bream['Length1'], bream['Weight'], label='bream')
plt.scatter(smelt['Length1'], smelt['Weight'], label='smelt')
plt.scatter(30,600, c='r', label='predict')
plt.legend()
plt.show()
````

*   **이웃 찾아내기**

````python
distance,indexes=kn.kneighbors([[30,600]])
print(distance, indexes)
````

````python
fish.iloc[indexes[0]]['Length1']
````

````python
plt.scatter(bream['Length1'], bream['Weight'], label='bream')
plt.scatter(smelt['Length1'], smelt['Weight'], label='smelt')
plt.scatter(30,600, c='r', label='predict')
plt.scatter(fish.iloc[indexes[0]]['Length1'], fish.iloc[indexes[0]]['Weight'], label='neighbors')
plt.legend()
plt.show()
````

*   <font color='red'>이웃의 개수를 늘린다면?</font>

````python
kn49=KNeighborsClassifier(n_neighbors=49)
kn49.fit(fish_x, fish_y)
print(kn49.score(fish_x, fish_y))
````

````python
for n in range(1, 50):
  kn_n=KNeighborsClassifier(n_neighbors=n)
  kn_n.fit(fish_x, fish_y)
  score=kn_n.score(fish_x, fish_y)

  if score<1:
    print(n, score)
    break
````
